### Python String Metotları Özet Tablosu

| Kategori | Fonksiyon | İşlevi | Örnek Girdi $\rightarrow$ **Çıktı** |
| --- | --- | --- | --- |
| **Büyük / Küçük Harf** | `.upper()` | Tüm harfleri büyük yapar. | `"atgc".upper()` $\rightarrow$ **`"ATGC"`** |
|  | `.lower()` | Tüm harfleri küçük yapar. | `"ATGC".lower()` $\rightarrow$ **`"atgc"`** |
|  | `.capitalize()` | Sadece **ilk** harfi büyük, kalanını küçük yapar. | `"atgc".capitalize()` $\rightarrow$ **`"Atgc"`** |
|  | `.title()` | Her kelimenin ilk harfini büyük yapar. | `"hasta adi".title()` $\rightarrow$ **`"Hasta Adi"`** |
| **Arama & Sayma** | `.count(x)` | Metin içinde `x` ifadesinin kaç kez geçtiğini sayar. | `"ATGCG".count("G")` $\rightarrow$ **`2`** |
|  | `.find(x)` | `x` ifadesinin ilk görüldüğü **indeksi (konumu)** verir. (Bulamazsa `-1` döner) | `"ATGCG".find("T")` $\rightarrow$ **`1`** |
|  | `.index(x)` | `.find()` ile aynıdır ama bulamazsa hata (`ValueError`) fırlatır. | `"ATGCG".index("C")` $\rightarrow$ **`3`** |
| **Kontrol (Sorgulama)** | `.startswith(x)` | Metin `x` ile başlıyor mu? (`True`/`False`) | `"AUGCC".startswith("AUG")` $\rightarrow$ **`True`** |
|  | `.endswith(x)` | Metin `x` ile bitiyor mu? (`True`/`False`) | `"UAG".endswith("AG")` $\rightarrow$ **`True`** |
|  | `.isalpha()` | Tüm karakterler **harf** mi? | `"ATGC".isalpha()` $\rightarrow$ **`True`** |
|  | `.isdigit()` | Tüm karakterler **sayı** mı? | `"123".isdigit()` $\rightarrow$ **`True`** |
| **Düzenleme & Temizlik** | `.replace(x, y)` | `x` karakterlerini `y` ile değiştirir. | `"ATGC".replace("T", "U")` $\rightarrow$ **`"AUGC"`** |
|  | `.strip()` | Metnin başındaki ve sonundaki tüm **boşlukları** temizler. | `"  ATGC  ".strip()` $\rightarrow$ **`"ATGC"`** |
|  | `.split(x)` | Metni verilen `x` karakterine göre **bölüp liste** yapar. | `"A,T,G,C".split(",")` $\rightarrow$ **`['A','T','G','C']`** |
|  | `.join(liste)` | Bir listedeki metinleri tek bir metin halinde **birleştirir**. | `"-".join(['A','T','G'])` $\rightarrow$ **`"A-T-G"`** |

## Parantez İçi GENELDE BOŞ Bırakılan Fonksiyonlar

Bu fonksiyonlar doğrudan tüm metin üzerinde genel bir kural uyguladıkları için ekstra bir bilgiye ihtiyaç duymazlar.

* **`.upper()`**: Hepsini büyük yap (Ekstra bilgiye gerek yok $\rightarrow$ `dna.upper()`).
* **`.lower()`**: Hepsini küçük yap (Ekstra bilgiye gerek yok $\rightarrow$ `dna.lower()`).
* **`.strip()`**: Baştaki ve sondaki boşlukları sil (İçi boş bırakılırsa boşlukları siler $\rightarrow$ `dna.strip()`).
* **`.isalpha()`**: Hepsi harf mi? (Soru doğrudan metnin kendisine sorulur $\rightarrow$ `dna.isalpha()`).

---

### Özet Mantık Tablosu 💡

| Fonksiyon | Parantez İçi Ne İster? | Mantık / Neden? |
| --- | --- | --- |
| `dna.upper()` | **BOŞ** | Çünkü "hepsini büyüt" emri nettir, ekstra seçeneğe gerek yoktur. |
| `dna.count("A")` | **Metin / Harf** | Çünkü Python'ın neyi sayacağını bilmesi gerekir. |
| `dna.replace("T", "U")` | **İki Metin** | Çünkü neyin gidip yerine neyin geleceğini bilmesi gerekir. |
| `dna.split("-")` | **Ayrıştırıcı Sembol** | Nereden makas atıp böleceğini bilmesi gerekir. |

Python'da `.sort()` fonksiyonu **sadece Listeler (`list`) üzerinde çalışır. Metinler (`string`) üzerinde doğrudan ÇALIŞMAZ.**

Eğer elindeki veri seti bir **metin (string)** ise ve sen onun arkasına `.sort()` eklediysen, Python sana şu hatayı fırlatır:
`AttributeError: 'str' object has no attribute 'sort'` *(Metin objesinin sort diye bir özelliği yoktur!)*

Gelin bu sorunu nasıl çözeceğimizi ve arka plandaki mantığı adım adım görelim:

---

## 1. Neden Metinlerde `.sort()` Çalışmaz?

Python'da metinler (string) **değiştirilemez (immutable)** yapılardır. `.sort()` fonksiyonu ise uygulandığı veriyi **doğrudan yerinde kalıcı olarak sıralamaya** çalışır. Python metinlerin içindeki harflerin yerini doğrudan değiştirmene izin vermediği için hata verir.

---

## 2. Çözüm Yolları (Metni veya Listeyi Nasıl Sıralarsın?)

İhtiyacına göre 2 farklı çözüm yolun var:

### A) Eğer Elindeki Bir Liste (`list`) İse:

Eğer verin bir **liste** ise `.sort()` tıkır tıkır çalışır:

```python
# Alfabetik veya sayısal liste
genler = ["TP53", "BRCA1", "EGFR"]

genler.sort() # Listeyi yerinde alfabetik sıralar
print(genler) 

# Çıktı: ['BRCA1', 'EGFR', 'TP53']

```

---

### B) Eğer Elindeki Bir Metin (`string`) İse (Örn: Bir DNA Dizisi):

Bir metindeki harfleri (örneğin `"ATGC"` dizisini) alfabetik sıraya dizmek istiyorsan `.sort()` yerine **`sorted()`** fonksiyonunu kullanmalısın!

`sorted()` fonksiyonu metni alır, harf harf parçalayıp sıralı bir **liste** yapar. Sonra onu tekrar metin yapmak için geçen ders öğrendiğimiz **`.join()`** ile birleştiririz:

```python
dna = "ATGCGA"

# 1. Yöntem: sorted() kullanımı (Sana sıralı bir harf listesi verir)
sirali_liste = sorted(dna)
print(sirali_liste)  
# Çıktı: ['A', 'A', 'C', 'G', 'G', 'T'] (Alfabetik sıraya girdi!)

# 2. Yöntem: O listeyi tekrar tek bir DNA metni yapmak istersen (join ile):
sirali_dna = "".join(sorted(dna))
print(sirali_dna)    
# Çıktı: "AACGGT"

```

---

## Özetle Fark Nedir? 💡

| Kullanım | Hangi Veri Tipi İçin? | Nasıl Yazılır? |
| --- | --- | --- |
| **`.sort()`** | **Sadece Listeler** (`list`) | `liste.sort()` *(Listeyi doğrudan kalıcı olarak değiştirir)* |
| **`sorted()`** | **Her Şey** (Metin, Liste, Tuple...) | `sorted(metin)` *(Orjinal veriyi bozmaz, sıralı yeni bir liste verir)* |


## Parantez Rehberi

* **`[1, 2, 3]` (Köşeli Parantez):** **`List` (Liste)** demektir. Esnektir, değiştirilebilir. `.sort()` **çalışır**.
* **`(1, 2, 3)` (Normal Parantez):** **`Tuple` (Demet)** demektir. Sabittir, kilitlidir. `.sort()` **çalışmaz** (Hata verir), onun yerine `sorted()` kullanılır.

Olay tamamen hangi parantezi kullandığınla ilgiliymış! Parantezi `[ ]` olarak değiştirip denersen sonucun hemen sıralandığını göreceksin.

## 1. Tersten (Büyükten Küçüğe) Sıralama: `reverse=True`

Varsayılan olarak Python her şeyi küçükten büyüğe (veya A'dan Z'ye) sıralar. Eğer en yüksek değerden en düşüğe sıralamak istersen parantez içine `reverse=True` parametresini eklersin.

### Sayılarda (Örn: Tümör Boyutları):

```python
tumor_boyutlari = [1.2, 4.5, 2.8, 0.9, 3.1]

# Büyükten küçüğe sıralayalım
tumor_boyutlari.sort(reverse=True)
print(tumor_boyutlari)  
# Çıktı: [4.5, 3.1, 2.8, 1.2, 0.9]

```

### Metinlerde (Z'den A'ya Alfabetik):

```python

sirali_genler = sorted(genler, reverse=True)
print(sirali_genler)  
# Çıktı: ['TP53', 'TNF', 'EGFR', 'BRCA1']

```

## 2. Özel Kurala Göre Sıralama: `key=` Kullanımı

Sıralamayı standart alfabetik veya sayısal sıraya göre değil, **kendi belirlediğin özel bir kurala göre** yaptırabilirsin. Bunun için `key=` parametresi kullanılır.

### A) Uzunluğa Göre Sıralama (`key=len`)

Metinleri alfabetik sıraya göre değil, **içindeki harf/baz sayısına göre (kısadan uzuna)** sıralamak istersen:

```python
dna_dizileri = ["ATGCGATCGAT", "ATG", "ATGCGA", "AT"]

# Harf uzunluğuna (len) göre sırala
dna_dizileri.sort(key=len)
print(dna_dizileri)  
# Çıktı: ['AT', 'ATG', 'ATGCGA', 'ATGCGATCGAT'] (En kısadan en uzuna dizildi!)

```

---

### B) Büyük-Küçük Harf Duyarlılığını Yok Etme (`key=str.lower`)

Python varsayılan olarak **büyük harfleri her zaman küçük harflerden önce** sıralar. Bu durum biyolojide veri karmaşasına yol açabilir:

```python
genler = ["tp53", "BRCA1", "egfr", "TNF"]

# Standart sıralama yaparsak (Büyük harfler öne geçer):
print(sorted(genler))  
# Çıktı: ['BRCA1', 'TNF', 'egfr', 'tp53']  <-- Harf büyüklüğü sırayı bozdu!

# Harf büyüklüğünü göz önünde bulundurmadan sıralamak için:
print(sorted(genler, key=str.lower))  
# Çıktı: ['BRCA1', 'egfr', 'TNF', 'tp53']  <-- Gerçek alfabetik sıra!

```

## 3. Karmaşık / Çiftli Verileri Sıralama (Tuple & Listeler)

Biyoistatistikte veriler genelde ikililer halinde gelir (Örn: `(Hasta_ID, Yaş)` veya `(Gen_Adı, İfade_Seviyesi)`).

Python bu tarz ikili verilerde **otomatik olarak ilk elemana göre** sıralama yapar:

```python
# (Gen Adı, İfade Seviyesi)
gen_verileri = [("TP53", 2.4), ("BRCA1", 5.1), ("EGFR", 1.8)]

# İlk elemana (Gen Adına) göre A'dan Z'ye sıralar:
gen_verileri.sort()
print(gen_verileri)  
# Çıktı: [('BRCA1', 5.1), ('EGFR', 1.8), ('TP53', 2.4)]

```

### İkinci Elemana (Sayısal Değere) Göre Sıralamak İstersek?

İşte burada `lambda` denilen küçük minik fonksiyonlar kullanılır. *"2. elemana (sayıya) göre sırala"* demek için:

```python
# 2. elemana (yani indeks 1 olan İfade Seviyesine) göre küçükten büyüğe sırala:
gen_verileri.sort(key=lambda x: x[1])
print(gen_verileri)  
# Çıktı: [('EGFR', 1.8), ('TP53', 2.4), ('BRCA1', 5.1)]

```

## 1. Eleman Silme ve Temizleme Araçları

Elindeki bir listeden eleman veya eleman grubu çıkartmak için kullanılan araçlardır.

### A) `.remove(deger)` $\rightarrow$ Değere Göre Siler

* **Ne işe yarar?** Listede gördüğü **ilk** belirtilen değeri siler. Konum (indeks) bilmene gerek yoktur, doğrudan silmek istediğin veriyi yazarsın.
* **Koşul:** Eğer silmek istediğin değer listede yoksa Python `ValueError` hatası verir.

```python
genler = ["BRCA1", "TP53", "EGFR", "TP53"]
genler.remove("TP53") # Sadece İLK gördüğü TP53'ü siler
print(genler) # Çıktı: ['BRCA1', 'EGFR', 'TP53']

```

### B) `.pop(indeks)` $\rightarrow$ Konuma Göre Siler ve Sildiği Elemanı Getirir

* **Ne işe yarar?** Belirtilen indisteki elemanı listeden söküp atar. Parantez içini **boş bırakırsan** en sondaki elemanı siler.
* **Farkı:** Sildiği elemanı sana geri döndürür, böylece sildiğin veriyi başka bir değişkene atayıp saklayabilirsin.

```python
kodonlar = ["AUG", "CGA", "UGA"]
son_kodon = kodonlar.pop() # En sondaki "UGA"'yı sildi ve hafızaya aldı
print(kodonlar)   # Çıktı: ['AUG', 'CGA']
print(son_kodon)  # Çıktı: 'UGA'

```

### C) `del` (Delete) $\rightarrow$ Genel Silme Anahtarı (Komut)

* **Ne işe yarar?** Bir liste fonksiyonu değildir, Python'ın öz anahtar kelimesidir (`keyword`). Indeks belirterek tek bir elemanı, dilimleyerek (`[1:3]`) bir aralığı veya değişkenin kendisini hafızadan tamamen silmek için kullanılır.

```python
hasta_id = [101, 102, 103, 104, 105]

del hasta_id[0]      # 0. indisteki 101'i siler -> [102, 103, 104, 105]
del hasta_id[1:3]    # 1. ve 2. indistekileri topluca siler -> [102, 105]

```

### D) `.clear()` $\rightarrow$ İçini Tamamen Boşaltır

* **Ne işe yarar?** Listeyi silmez ama içindeki tüm elemanları temizleyip içi boş bir liste (`[]`) bırakır.

```python
numuneler = ["A1", "B2", "C3"]
numuneler.clear()
print(numuneler) # Çıktı: []

```

---

## 2. Arama ve Varlık Kontrol Araçları

Listede bir elemanın olup olmadığını test etmek veya konumunu bulmak için kullanılır.

### A) `in` ve `not in` $\rightarrow$ Var mı / Yok mu Kontrolü

* **Ne işe yarar?** Bir elemanın listenin veya metnin içinde olup olmadığını kontrol eder. Cevap olarak doğrudan `True` veya `False` döner.

```python
dna = "ATGCGAT"
print("ATG" in dna)      # Çıktı: True
print("U" not in dna)    # Çıktı: True

```

### B) `.index(deger)` $\rightarrow$ Aranan Elemanın Yerini (İndeksini) Bulur

* **Ne işe yarar?** Bir elemanın liste içinde kaçıncı sırada (indekste) olduğunu söyler.

```python
amino_asitler = ["Met", "Leu", "Ala"]
konum = amino_asitler.index("Leu")
print(konum) # Çıktı: 1

```

---

## 3. Eleman Ekleme Araçları

Var olan bir listeye yeni veriler eklemek için kullanılan araçlardır.

### A) `.append(deger)` $\rightarrow$ En Sona Ekler

* **Ne işe yarar?** Veriyi listenin **en sonuna** yapıştırır.

```python
genler = ["BRCA1"]
genler.append("TP53") # ['BRCA1', 'TP53']

```

### B) `.insert(indeks, deger)` $\rightarrow$ Araya / İstenen Konuma Ekler

* **Ne işe yarar?** Belirttiğin indeks sırasına yeni elemanı yerleştirir, oradaki eski elemanları sağa kaydırır.

```python
kodonlar = ["AUG", "UGA"]
kodonlar.insert(1, "CGA") # 1. sıraya "CGA" yerleştir
print(kodonlar) # Çıktı: ['AUG', 'CGA', 'UGA']

```

### C) `.extend(liste)` $\rightarrow$ İki Listeyi Birleştirir

* **Ne işe yarar?** Bir listenin sonuna başka bir listenin tüm elemanlarını topluca ekler.

```python
genler_set1 = ["BRCA1", "TP53"]
genler_set2 = ["EGFR", "TNF"]
genler_set1.extend(genler_set2)
print(genler_set1) # Çıktı: ['BRCA1', 'TP53', 'EGFR', 'TNF']

```

## Özet Karşılaştırma Tablosu 💡

| Araç / Fonksiyon | Neye Göre İş Yapar? | Ne Yapar? |
| --- | --- | --- |
| **`.remove("x")`** | Değere göre | İlk gördüğü `"x"` değerini listeden siler. |
| **`.pop(0)`** | Konuma (İndekse) göre | Belirtilen indeksteki elemanı listeden **söker ve getirir**. |
| **`del liste[0]`** | Konuma veya Aralığa göre | Belirtilen indeksi veya aralığı hafızadan siler. |
| **`"x" in liste`** | Değere göre | Eleman var mı diye bakar (`True` / `False`). |
| **`.append("x")`** | En sona | Yeni elemanı en sona ekler. |
| **`.insert(0, "x")`** | Belirtilen konuma | Elemanı istenen indekse araya ekler. |
